```python
class Profile:
  def __init__(age_range, gender, occupation, things_of_importance, current_commitment):
    self.name = ""
    self.age_range = age_range
    self.gender = gender
    self.occupation = occupation
    self.marital_status = ''
    self.achiever_level = 1
    self.things_of_importance = things_of_importance
    self.current_commitment = current_commitment

  def showProfile(self):
    print(self)
```
